```python
skill=input("enter skill: ")
exp=int(input("enter exp: "))
loc=input("enter location: ")
if skill=="python":
    if exp==1:
        if loc=="cbe":
            print("BESANT")
        else:
            print("INVALID LOCATION")
    elif exp>=2:
        if loc=="chn":
            print("ZOHO")
        else:
            print("INVALID LOCATION")
    else:
        print("INVALID EXPERIANCE")

elif skill=="java":
    if exp<2:
        if loc=="cbe":
            print("TCS")
        else:
            print("INVALID LOCATION")
    elif exp>=4:
        if loc=="chn":
            print("CTS")
        else:
            print("INVALID LOCATION")
    else:
        print("INVALID EXPERIANCE")

else:
    print("INVALID SKILL")      
```
