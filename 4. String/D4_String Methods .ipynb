# String methods

## 1. Lower

#### changes the given string to full lower case letters

```python
a= " My Name Is Dhivyaa " #changes the given text to lower case
print(a.lower())
```

```python
b = "hello 123"    #checks whether the given string is in lower case or not
print(b.islower())
```

```python
a = "Hello world!"
print(a.islower())
```

## 2. Casefold

#### similar to lower() but casefold() stronger 
#### even converts the german and other languages into lower case where as lower dont

```python
b = "Straße"
print(b.lower())
print(b.casefold())
```

```python
a = "maß"

print(a.lower())
print(a.casefold())
```

## 3. Upper 

#### converts everything to uppercase

```python
text = "Hello Python" # change string to python 
print(text.upper())
```

## 4. Swapcase

####  uppercase becomes lowercase and lowercase becomes uppercase

```python
text = "Hello Python"
print(text.swapcase())
```

## 5. Strip

#### removes spaces from both sides

```python
text = "   The children are playing in ground   "
print(text.strip())
```

## 6. lstrip

#### removes spaces from left side only

```python
d= "       The children are playing in ground   "
print("|"+ d.lstrip()+ "|")
print("|"+ d.strip()+ "|")
```

## 7. Replace

#### replaces a specific string

```python
e = "Hello World"
print(e.replace("World", "Everyone"))
```

## 8. isalpha

#### returns True if all the characters are alphabet letters (a-z)

```python
f="chennai10"
print(f.isalpha())
```

```python
f="pondicherry"
print(f.isalpha())
```

## 9. Find

#### method finds the first occurrence of the specified value ( PROVIDE THE INDEX VALUE OF THE FIRST LETTER FROM WHERE THE REQUIRED WORD STARTS )
#### returns -1 if the value is not found

```python
g = "Today is Thursday"
print(g.find("Thursday"))
```

```python
g = "Today is Thursday"
print(g.find("Dog"))
```

## 10. Capitalize

#### changes first character in string to upper case, and the rest to lower case.

```python
h = "python is FUN!"
print(h.capitalize())
```

```python
i = "22 is my age."
print(i.capitalize())
 
```

## 11. Title

```python
text = "hello python programming"

print(text.capitalize())
print(text.title())
```

## 12. Replace

#### replaces the specific word with another new word as provided

```python
j= " she was playing cards with her friends"
print(j.replace("cards","shuttle"))
```
