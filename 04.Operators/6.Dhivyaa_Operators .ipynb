#  OPERATORS

## 1. ARITHEMATIC OPERATOR

```python
a=11
b=9
print("ADD:" ,a+b)
print("SUBTRACT:",a-b)
print("MULTIPLY:",a*b)
print("DIVISION:",a/b)
print("FLOOR DIVISION:",a//b)
print("MODULUS:",a%b)
```

## 2. ASSIGNMENT OPERATOR

```python
a=116
a+=5
print("add and assign=",a)     #116+5=121
a-=6
print("subtract and assign=",a) #121-6=115
a*=2
print("multiply and assign=",a) #115x2=230
a/=4
print("divide and assign=",a) #230/4=57.5
a//=4
print("floor division and assign=",a) #57.5//4=14
a%=4
print("mod and assign=",a)  #14 % 4=2
a**=3
print("exponent and assign=",a) #2^3=8
```

## 3. COMPARISON OPERATOR

```python
a=116
b=96
print("equal to: ",a==b)
print("not equal to: ",a!=b)
print("greater than or equal to: ",a>=b)
print("lesser than or equal to: ",a<=b)
print("greater than: ",a>b)
print("lesser than: ",a<b)
```

## 4. LOGICAL OPERATOR

#### AND ---> both condition must be true

```python
a=243
b=143
print(a>b and b<a) # both condition pass
(a<b and b<a) # one condition only pass
```

```python
a="chennai"
b="pondicherry"
print(a>b and b<a)
```

```python
ord ("c")
```

```python
ord("p")
```

#### OR ---> any one condition must be true

```python
a=100
b=50
print(a>50 or b<75)
print(a>150 or b<75) #one condition pass
```

#### NOT ---> reverse the result

```python
a=11
print(not(a>5))
```

## 6. MEMBERSHIP OPERATOR

#### IN --->check whether value is present

```python
#list
a=[1,2,3]
print(2 in a)
(5 in a)
```

```python
#string
b="dhivyaa"
print("y" in b)
("s" in b)
```

```python
e = " all is well"
print("well" in e)
("good" in e)
```

#### NOT IN---> check whether value is absent

```python
d= ["chennai","bangalore","pondicherry"]
print("pondicherry" not in d)
("ooty" not in d)
```

```python
a=[12,24,36]
print(24 not in a)
(5 not in a)
```

## 7. BITWISE OPERATOR

```python
# AND(&)
a=34
b=14
print(a&b)
```

```python
# OR(|)
a=8
b=9
print(a|b)
```

```python
# XOR(^)
a=16
b=20
print(a^b)
```

```python
# NOT(~)
A=116
print(~A)
```

```python
# LEFT SHIFT(<<)
a=17
print("one shift: ",a<<1)
print("two shift: ",a<<2)
print("three shift: ",a<<3)
print("four shift: ",a<<4)
```

```python
# RIGHT SHIFT(>>)
a=16
print("one shift: ",a>>1)
print("two shift: ",a>>2)
print("three shift: ",a>>3)
print("four shift: ",a>>4)
```
