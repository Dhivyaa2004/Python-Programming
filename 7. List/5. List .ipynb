## change list items

```python
a=["apple","banana","carrot"]
a.insert(2,"kiwi")
print(a)
```

```python
a=["apple","banana","carrot"] # replace isnt a keyword in list
a.replace(2,"kiwi")
print(a)
```

```python
a=["apple","banana","carrot"]
a[1]="kiwi"
a
```

```python
a=["apple","banana","carrot"] #replaces apple with new fruits
a[:1]=["kiwi","mango"]
a
```

```python
k=[1,2,33,34,67]
k.append(44)
k
```

# removing of elements - pop()

```python
a=["apple","banana","carrot"] 
a.remove(1)
a
```

```python
a=["apple","banana","carrot"] # for replace use value inside
a.remove("banana")
a
```

```python
a=["apple","banana","carrot"] #removes last term
a.pop()
a
```

```python
a=["apple","banana","carrot"] # for pop use index number inside
a.pop(1)
a
```

```python
a=["apple","banana","carrot"] 
del a[0]
a
```

```python
a=["apple","banana","carrot","avacado","plum","cherry"] #chnages the list to ascending order
a.sort()
a
```

```python
a=["apple","banana","carrot","avacado","plum","cherry"] #just reverses the string
a.reverse()
a
```

```python
a=["apple","banana","carrot","avacado","plum","cherry"] # reverses the string by sorting it in ascending order
a.sort(reverse=True)
a
```
