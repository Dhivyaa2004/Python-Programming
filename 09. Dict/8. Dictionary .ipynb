# delete

## pop


```python
a={"brand":"ford","model":"mustang","year":1964,"brand1":"ford"}
a.pop("brand") #deletes the specified key 
a
```

```python
## pop item
```

```python
a={"brand":"ford","model":"mustang","year":1964,"brand1":"ford"}
a.popitem() #deletes the last key 
a
```

# NESTED DICTIONARY

```python
child1={"name":"emil","year":2004}
child2={"name":"abc","year":2002,"class":90}
child1.update(child2)
new={"native":"chennai","height":100}
child1.update(new)
child1
```

```python
child1={"name":"emil","year":2004}
child1["name"]="abc"
child1["year"]=2002
child1.update({"class":90})
new={"native":"chennai","height":100}
child1.update(new)
child1
```
