# Assignment Operator

```python
# Assignment Operators

a = 10
print(a) # directly assign value to variable

a = 10
a += 5
print(a) # add and provides the assigned value

a = 10
a -= 5
print(a) # subtracts and provides the assigned value

a = 10
a *= 5
print(a) # multiply and provides the assigned value

a = 10
a /= 5
print(a) # divide and provides the assigned value

a = 10
a //= 3
print(a)  # floor divison and provides the assigned value

a = 10
a %= 3
print(a) # mod and provides the assigned value

a = 2
a **= 3
print(a) #Exponentiation assignment | Raises a number to a power and assigns the result

```

```python
# Extra Asssignment Operator

a = 10
a &= 3
print(a) #Bitwise AND assignment | Performs bitwise AND and assigns the result

a = 10
a |= 3
print(a) #Bitwise OR assignment | Performs bitwise OR and assigns the result

a = 10
a ^= 3
print(a) # Bitwise XOR assignment | Performs bitwise XOR and assigns the result

a = 10
a >>= 1
print(a) #Right shift assignment | Shifts bits to the right and assigns the result

a = 10
a <<= 1
print(a) #Left shift assignment | Shifts bits to the left and assigns the result
```
