# Logical Operator

### AND - both conditions must be true

```python
a=10
b=5
print(a>b and b<a)
```

```python
print (a>50 and b>4) # although b>4 condition passes but a >50 fails so output turns to be false
```

### OR - either one condition should be true

```python
a=10
b=5
print(a>b or b<a)  #both condition passes
```

```python
print (a>50 or b>4) # one condition passes
```

```python
print (a>50 or b>10) # both condition fails
```

### NOT - reverses the result

```python
b=5
print(not(b>10)) # 5 is not greater than 10 so the output must be false since we are using NOT it reverses the outut and provide True
```

```python
b=5
print(not(b>3))
```

# BITWISE OPERATOR

### AND(&) - using binary numbers the calulation is done - returns 1 when both bits are 1


```python
a=10         # 1010
b=3          # 0011 ----> 0010 = 2
print(a&b)
```

### OR( | ) - using binary numbers the calulation is done - returns 1 when either any one bit is 1

```python
a=10         # 1010
b=3          # 0011 ----> 1011 = 11
print(a|b)
```

### XOR(^) - using binary numbers the calulation is done - returns 1 when both bits are DIFFRENT

```python
a=10         # 1010
b=3          # 0011 ----> 1001 = 9
print(a^b)
```

### NOT (~) - Reverses bits USES FORMULA ------->  ~a=-(a+1)

```python
a=3
print(~a)   #~a = -(a+1)
```

### LEFT SHIFT - shifts the binary bits of a number to the LEFT by a specified number of positions.

```python
a=10        
print(a<<1)  # the bit in the opertions shifts by 1 place on the left side and the empty spaces on right side are given with default 0
(a<<2)       # the bit in the opertions shifts by 2 place on the left side and the empty spaces on right side are given with default 0
```

### RIGHT SHIFT - shifts the binary bits of a number to the right by a specified number of positions.

```python
a=10
print(a>>1) # the bits in the operation are shifted the 1 place on right side and there is not any place to shift they are removed
print(a>>2)
(a>>3)
```
